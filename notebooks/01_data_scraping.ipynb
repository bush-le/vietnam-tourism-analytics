# 01. Data Scraping & Collection - Vietnam Tourism Analytics

Notebook này thực hiện thu thập dữ liệu thống kê du lịch Việt Nam từ 2 nguồn chính thống:
- **Tổng cục Thống kê (GSO / NSO)**: https://www.nso.gov.vn/thuong-mai-dich-vu/ (PX-Web du lịch, chi tiêu du khách, cơ cấu doanh thu)
- **Cục Du lịch Quốc gia Việt Nam (VNAT)**: https://vietnamtourism.gov.vn/statistic (Khách quốc tế, khách nội địa, thị trường gửi khách hàng đầu, cơ sở lưu trú và buồng phòng)

```python
import os
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup

RAW_DIR = "../data/raw"
os.makedirs(RAW_DIR, exist_ok=True)
print(f"Thư mục lưu trữ raw data: {RAW_DIR}")
```

## 1. Kiểm tra các tệp dữ liệu đã thu thập trong data/raw/

```python
files = os.listdir(RAW_DIR)
for f in files:
    if f.endswith(".csv"):
        df = pd.read_csv(os.path.join(RAW_DIR, f))
        print(f"{f}: {df.shape[0]} dòng, {df.shape[1]} cột")
```
